```python
totalcolN2 = 0
totalcolO2 = 0
inel = 0
el = 0
for collision in diags.collisionRecorder.collisions:
    if collision.specie == "N2":
        totalcolN2 += 1
    elif collision.specie == "O2":
        totalcolO2 += 1

    if collision.type == "inelastic":
        inel += 1
    elif collision.type == "elastic":
        el += 1

print(f"Total collisions with N2: {totalcolN2}")
print(f"Total collisions with O2: {totalcolO2}")
print(f"Total inelastic collisions: {inel}")
print(f"Total elastic collisions: {el}")
```
