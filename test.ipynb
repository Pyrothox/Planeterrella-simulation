```python
import tomllib
import numpy as np
with open("config.toml", "rb") as f:
    data = tomllib.load(f)

print(data)

gas_params = data["Gas"]
species = gas_params["species"]
names = np.array([specie["name"] for specie in species])
fractions = np.array([specie["fraction"] for specie in species])
print(fractions)
```

```python
import pyvista as pv

sphere1 = pv.Sphere(radius=data["SmallSphere"]["R"], center=data["SmallSphere"]["X"])
sphere2 = pv.Sphere(radius=data["LargeSphere"]["R"], center=data["LargeSphere"]["X"])

plotter = pv.Plotter()
plotter.set_background("black")

plotter.add_mesh(sphere1, color="silver")
plotter.add_mesh(sphere2, color="silver")
plotter.show()
```
